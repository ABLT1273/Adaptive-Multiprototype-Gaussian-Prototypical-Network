```python
from amgpn.config import require as _cfg_require, resolve as _cfg_resolve, format_value as _cfg_format
import os

# ----------------- 定义参数 -----------------
DATASET_ID = "kitofrank/RFUAV"
# 关键：指定截图中的分支/引用
REVISION = "'refs/convert/parquet'" 
# 定义本地保存的文件夹名称
LOCAL_DIR = _cfg_require('GPN_UAV_data/downloadDataset.ipynb.cell1.module.path') 

# ----------------- 检查并创建目录 -----------------
if not os.path.exists(LOCAL_DIR):
    os.makedirs(LOCAL_DIR)
    print(f"已创建目录: {LOCAL_DIR}")
else:
    print(f"目录 {LOCAL_DIR} 已存在，文件将被下载到其中。")

# ----------------- 构建并执行命令行 -----------------
# 注意：将 repo-id 直接放在 download 后面，并删除 --repo-id 旗标
command = (
    f"huggingface-cli download "
    f"{DATASET_ID} "
    f"--revision {REVISION} "
    f"--local-dir {LOCAL_DIR} "
    f"--local-dir-use-symlinks False "
    f"--repo-type dataset" 
)
print("执行修复后的下载命令...")
print(f"命令: {command}")

# 在 Jupyter Notebook 或 Colab 中运行命令行
!{command}

print("\n下载完成！")
print(f"数据集已保存到本地文件夹: {LOCAL_DIR}")
```
