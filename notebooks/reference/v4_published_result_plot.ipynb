# Published-reference result plot
Plots stored reference numbers, not newly measured model results. Keep this separate from evaluation workflows.


```python
from amgpn.config import require as _cfg_require, resolve as _cfg_resolve, format_value as _cfg_format
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
"""配置符合学术出版标准的matplotlib样式"""
plt.style.use('seaborn-v0_8-paper') # 使用适合论文的预设

# 字体配置 (推荐 Times New Roman 或 Arial)
plt.rcParams.update({
    'font.family': 'serif',
    'font.serif': ['Times New Roman', 'Times', 'DejaVu Serif'],
    'axes.labelsize': 12,
    'font.size': 12,
    'legend.fontsize': 10,
    'xtick.labelsize': 10,
    'ytick.labelsize': 10,
    'text.usetex': False, # 如果系统安装了LaTeX可设为True，否则False
    'figure.dpi': _cfg_require('GPN_V4_test.ipynb.cell1.module.size_or_budget'),
    'savefig.dpi': _cfg_require('GPN_V4_test.ipynb.cell1.module.size_or_budget__2'),
    'axes.grid': True,
    'grid.alpha': 0.3,
    'grid.linestyle': '--',
    'errorbar.capsize': 4,      # cap的宽度
    'lines.markeredgewidth': 1.5  # cap的线条粗细
})
palette = sns.color_palette("colorblind", 10)

colors = {
    'AMGPN': palette[1],    # 黄色
    'GPN': palette[9],      # 绿色
    'ProtoNet': palette[8], # 红色
    'MN': palette[0]        # 蓝色
}

# 数据
datasets = {
    'DroneRFb-Spectra': {
        'MN': {'acc': [72.7, 79.1, 80.8, 82.0], 'ci': [1.9, 1.6, 1.6, 1.5]},
        'ProtoNet': {'acc': [82.7, 88.9, 91.7, 90.5], 'ci': [2.0, 1.3, 0.9, 0.9]},
        'GPN': {'acc': [81.9, 90.0, 91.3, 92.0], 'ci': [2.0, 1.2, 1.0, 0.8]},
        'AMGPN': {'acc': [81.8, 93.7, 94.9, 96.3], 'ci': [2.0, 0.8, 0.6, 0.4]},

    },
    'RFUAV': {
        'MN': {'acc': [48.0, 45.3, 45.1, 44.7], 'ci': [2.0, 2.3, 1.7, 1.8]},
        'ProtoNet': {'acc': [56.2, 64.6, 65.9, 66.1], 'ci': [1.9, 1.7, 1.8, 1.7]},
        'GPN': {'acc': [55.5, 66.3, 69.6, 71.4], 'ci': [1.7, 1.6, 1.6, 1.4]},
        'AMGPN': {'acc': [51.7, 74.2, 81.6, 86.1], 'ci': [1.8, 1.6, 1.3, 1.3]},
    }
}

shots = _cfg_require('GPN_V4_test.ipynb.cell1.module.shots')
models = [ 'MN', 'ProtoNet','GPN', 'AMGPN',]

# 创建图形
fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=False)

for idx, (dataset_name, data) in enumerate(datasets.items()):
    ax = axes[idx]
    x = np.arange(len(shots))
    width = _cfg_require('GPN_V4_test.ipynb.cell1.module.width')
    
    for i, model in enumerate(models):
        offset = (i - 1.5) * width
        ax.bar(x + offset, 
               data[model]['acc'], 
               width, 
               yerr=data[model]['ci'],
               capsize=3,
               label=model if idx == 0 else "",  # 只在第一个子图显示图例
               color=colors[model],
               alpha=_cfg_require('GPN_V4_test.ipynb.cell1.module.alpha__2'),
               edgecolor='black',
               linewidth=0.8)
    
    # 设置标签和标题
    ax.set_xlabel('K-shot', fontsize=13, fontweight='bold')
    ax.set_ylabel('Accuracy (%)', fontsize=13, fontweight='bold')
    ax.set_title(dataset_name, fontsize=14, fontweight='bold', pad=15)
    ax.set_xticks(x)
    ax.set_xticklabels(shots)
    ax.tick_params(labelsize=11)
    ax.grid(axis='y', alpha=_cfg_require('GPN_V4_test.ipynb.cell1.module.alpha'), linestyle='--')
    
    # 设置Y轴范围
    if dataset_name == 'DroneRFb-Spectra':
        ax.set_ylim(70, 100)
    else:
        ax.set_ylim(40, 90)

# 添加统一图例
handles, labels = axes[0].get_legend_handles_labels()
fig.legend(handles, labels, 
           loc='upper center', 
           bbox_to_anchor=(0.5, 1.05),
           ncol=4, 
           fontsize=12,
           frameon=True,
           edgecolor='black')

# 添加总标题
fig.suptitle('Few-Shot Classification Performance (95% Confidence Intervals)', 
             fontsize=15, fontweight='bold', y=1.12)

plt.tight_layout()
plt.savefig('comparison_two_datasets.pdf', bbox_inches='tight', dpi=_cfg_require('GPN_V4_test.ipynb.cell1.module.size_or_budget__3'))
plt.show()
```
